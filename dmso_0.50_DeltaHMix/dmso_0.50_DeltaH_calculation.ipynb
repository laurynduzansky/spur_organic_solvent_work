```python
import pandas as pd

# Load the CSV file into a DataFrame
df_pure_water_050dmso = pd.read_csv('pure_water_0.50dmso_data.csv')
print (df_pure_water_050dmso.columns)
```

```python
import sys
print(sys.executable)
import pymbar
```

```python
from pymbar import timeseries

# uses pymbar to find equilibration values and stores them in a new array A_n

A_t = df_pure_water_050dmso["Total Energy (kJ/mole)"]

t0, g, Neff_max = timeseries.detect_equilibration(A_t) # compute indices of uncorrelated timeseries
print(f"t0: {t0}")
print(f"g: {g}")
print(f"Neff_max: {Neff_max}")
A_t_equil = A_t[t0:]
indices = timeseries.subsample_correlated_data(A_t_equil, g=g)
A_n = A_t_equil.iloc[indices]
```

```python
# takes the mean value of equilibrated data
equil_pure_water_050dmso_total_energy = A_n.mean()
print (equil_pure_water_050dmso_total_energy, "kJ/mol")
```

```python
A_t = df_pure_water_050dmso["Box Volume (nm^3)"]

t0, g, Neff_max = timeseries.detect_equilibration(A_t) # compute indices of uncorrelated timeseries
print(f"t0: {t0}")
print(f"g: {g}")
print(f"Neff_max: {Neff_max}")
A_t_equil = A_t[t0:]
indices = timeseries.subsample_correlated_data(A_t_equil, g=g)
A_n_v = A_t_equil.iloc[indices]
```

```python
# takes the mean value of equilibrated data
equil_pure_water_050dmso_total_volume = A_n_v.mean()
print (equil_pure_water_050dmso_total_volume, "nm^3")
```

```python
df_pure_dmso_050 = pd.read_csv('pure_dmso_0.50_data.csv')
print (df_pure_dmso_050.columns)
```

```python
A_t = df_pure_dmso_050["Total Energy (kJ/mole)"]

t0, g, Neff_max = timeseries.detect_equilibration(A_t) # compute indices of uncorrelated timeseries
print(f"t0: {t0}")
print(f"g: {g}")
print(f"Neff_max: {Neff_max}")
A_t_equil = A_t[t0:]
indices = timeseries.subsample_correlated_data(A_t_equil, g=g)
A_n3 = A_t_equil.iloc[indices]
```

```python
A_t = df_pure_dmso_050["Box Volume (nm^3)"]

t0, g, Neff_max = timeseries.detect_equilibration(A_t) # compute indices of uncorrelated timeseries
print(f"t0: {t0}")
print(f"g: {g}")
print(f"Neff_max: {Neff_max}")
A_t_equil = A_t[t0:]
indices = timeseries.subsample_correlated_data(A_t_equil, g=g)
A_n_v3 = A_t_equil.iloc[indices]
```

```python
equil_pure_dmso_050_total_volume = A_n_v3.mean()
print (equil_pure_dmso_050_total_volume, "nm^3")
```

```python
pure_water_050_volume_m3 = (equil_pure_water_050dmso_total_volume * 1e-27)
print (pure_water_050_volume_m3, "m^3")
```

```python
dmso_water_mix_050_volume_m3 = (equil_dmso_water_mix_050_total_volume * 1e-27)
print (dmso_water_mix_050_volume_m3, "m^3")
```

```python
pure_dmso_050_volume_m3 = (equil_pure_dmso_050_total_volume * 1e-27)
print (pure_dmso_050_volume_m3, "m^3")
```

```python
pressure_kPa = 101.325 
print (pressure_kPa, "kPa")
```

```python
H_pure_water_050dmso = (equil_pure_water_050dmso_total_energy + (pressure_kPa * pure_water_050_volume_m3))
print (H_pure_water_050dmso, "kJ/mole")
```

```python
H_dmso_water_mix_050 = (equil_dmso_water_mix_050_total_energy + (pressure_kPa * dmso_water_mix_050_volume_m3))
print (H_dmso_water_mix_050, "kJ/mole")
```

```python
H_pure_dmso_050 = (equil_pure_dmso_050_total_energy + (pressure_kPa * pure_dmso_050_volume_m3))
print (H_pure_dmso_050, "kJ/mole")
```

```python
change_H_mixing_dmso_water_050 = (H_dmso_water_mix_050 - ((0.50 * H_pure_water_050dmso)+(0.50 * H_pure_dmso_050)))
total_change_H_mixing_dmso_water_050 = (change_H_mixing_dmso_water_050 / 500)
print (total_change_H_mixing_dmso_water_050, "kJ/mole")
```
