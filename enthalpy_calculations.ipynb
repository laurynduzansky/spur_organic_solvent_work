```python
from pymbar import timeseries

# uses pymbar to find equilibration values and stores them in a new array A_n

A_t = df_filename["Total Energy (kJ/mole)"]

t0, g, Neff_max = timeseries.detect_equilibration(A_t) # compute indices of uncorrelated timeseries
print(f"t0: {t0}")
print(f"g: {g}")
print(f"Neff_max: {Neff_max}")
A_t_equil = A_t[t0:]
indices = timeseries.subsample_correlated_data(A_t_equil, g=g)
A_n = A_t_equil.iloc[indices]
```
